```python
import random
import pandas as pd
from IPython.display import display

# ตั้งค่า Seed เพื่อให้สุ่มผลลัพธ์เดิมทุกครั้งที่รัน
random.seed(42)

# รายชื่อและนามสกุลสำหรับใช้สุ่ม
first_names = ["ทองดี", "พิไล", "มุนตา", "เรยา", "มุนินทร์"]
last_names = ["วงศ์เสวต", "จงสวัสดิ์", "อัศวรุ่งเรืองกิจ", "ตากลี", "บุญพูลทรัพย์"]

# 1. ประกาศ Class ParcelStatusTracker
class ParcelStatusTracker:
    def __init__(self, intake_id, parcel_id, recipient_name, old_status):
        self.intake_id = intake_id
        self.parcel_id = parcel_id
        self.recipient_name = recipient_name
        self.old_status = old_status
        self.new_status = old_status

    def update_status(self, target_status):
        valid_statuses = ["กำลังส่ง", "ส่งสำเร็จ"]
        if target_status in valid_statuses:
            self.new_status = target_status

# 2. วนลูปสุ่มชื่อผู้รับจริง และสุ่มสถานะใหม่ 300 รายการ
status_options = ["กำลังส่ง", "ส่งสำเร็จ"]
records = []

for i in range(1, 301):
    intake_id = f"INT-{i:04d}"
    parcel_id = f"TH{i:05d}"

    # สุ่มชื่อ-นามสกุลจริง
    recipient_name = f"{random.choice(first_names)} {random.choice(last_names)}"
    old_status = "รับแล้ว"

    # สร้าง Object และสุ่มอัปเดตสถานะ
    tracker = ParcelStatusTracker(intake_id, parcel_id, recipient_name, old_status)
    next_status = random.choice(status_options)
    tracker.update_status(next_status)

    # บันทึกข้อมูล
    records.append({
        "intake_id": tracker.intake_id,
        "parcel_id": tracker.parcel_id,
        "recipient_name": tracker.recipient_name,
        "old_status": tracker.old_status,
        "updated_status": tracker.new_status
    })

# 3. แปลงเป็น DataFrame และ Export เป็นไฟล์ CSV
df_status = pd.DataFrame(records)
df_status.to_csv("parcel_status_updated.csv", index=False, encoding="utf-8-sig")

print("สร้างไฟล์ CSV พร้อมชื่อผู้รับจริงเรียบร้อยแล้ว: parcel_status_updated.csv")

# 4. สั่งกางแสดงผล 300 แถวเต็ม
pd.set_option('display.max_rows', None)
display(df_status)
```
